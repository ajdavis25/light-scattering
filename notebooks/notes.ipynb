**D. Monte Carlo Audit**
- Active Monte Carlo files/functions: [main.cpp#L16](c:/Users/ashton/Desktop/projects/light-scattering/monte_carlo_cpp/src/main.cpp#L16), [MonteCarloDriver.cpp#L53](c:/Users/ashton/Desktop/projects/light-scattering/monte_carlo_cpp/src/MonteCarloDriver.cpp#L53), [Atmosphere.cpp#L7](c:/Users/ashton/Desktop/projects/light-scattering/monte_carlo_cpp/src/Atmosphere.cpp#L7), [WavelengthHandling.cpp#L5](c:/Users/ashton/Desktop/projects/light-scattering/monte_carlo_cpp/src/WavelengthHandling.cpp#L5), [PhaseFunctions.cpp#L6](c:/Users/ashton/Desktop/projects/light-scattering/monte_carlo_cpp/src/PhaseFunctions.cpp#L6), [Polarization.cpp#L33](c:/Users/ashton/Desktop/projects/light-scattering/monte_carlo_cpp/src/Polarization.cpp#L33), and [SurfaceReflection.cpp#L10](c:/Users/ashton/Desktop/projects/light-scattering/monte_carlo_cpp/src/SurfaceReflection.cpp#L10). The older standalone MC is [spherical/MonteCarloSky.cpp#L260](c:/Users/ashton/Desktop/projects/light-scattering/spherical/MonteCarloSky.cpp#L260).
- How the current algorithm works: `runMonteCarloSimulation()` loops over wavelength bands, launches photons from TOA in `launchPhoton()`, samples a free path with `-log(u)/sigmaTot`, probabilistically absorbs or scatters, samples a new direction from a Rayleigh/HG mixture, bins exits in `recordExitLocal()`, then collapses all bins into one rough sRGB color in [MonteCarloDriver.cpp#L230](c:/Users/ashton/Desktop/projects/light-scattering/monte_carlo_cpp/src/MonteCarloDriver.cpp#L230).
- What is correct:
  - There is a real spherical shell geometry with Earth radius and TOA.
  - The code at least separates atmosphere, wavelengths, phase functions, polarization, and surface reflection into modules.
  - The free-path formula itself is the standard exponential form for a homogeneous medium.
  - The newer branch is clearly trying to move toward multi-wavelength color rather than a single-band toy model.
- What is suspicious or incorrect, programming bugs:
  - `launchPhoton()` launches from `[0,0,R_TOA]` with `baseTheta=85` and `dir.z=cos(theta)` at [MonteCarloDriver.cpp#L56](c:/Users/ashton/Desktop/projects/light-scattering/monte_carlo_cpp/src/MonteCarloDriver.cpp#L56), [MonteCarloDriver.cpp#L58](c:/Users/ashton/Desktop/projects/light-scattering/monte_carlo_cpp/src/MonteCarloDriver.cpp#L58), and [MonteCarloDriver.cpp#L68](c:/Users/ashton/Desktop/projects/light-scattering/monte_carlo_cpp/src/MonteCarloDriver.cpp#L68). In this coordinate system that points photons mostly sideways and slightly upward, not into the atmosphere. That alone can explain empty bins or black output.
  - The scattering update writes `p.dir = (sinT cos phi, sinT sin phi, mu)` in the global frame at [MonteCarloDriver.cpp#L141](c:/Users/ashton/Desktop/projects/light-scattering/monte_carlo_cpp/src/MonteCarloDriver.cpp#L141) and [MonteCarloDriver.cpp#L148](c:/Users/ashton/Desktop/projects/light-scattering/monte_carlo_cpp/src/MonteCarloDriver.cpp#L148). It does not rotate the sampled angle around the incoming ray. That makes the scattering kinematics wrong after the first event.
  - The rejection sampler uses `testVal = 4*u` at [MonteCarloDriver.cpp#L142](c:/Users/ashton/Desktop/projects/light-scattering/monte_carlo_cpp/src/MonteCarloDriver.cpp#L142), but [PhaseFunctions.cpp#L11](c:/Users/ashton/Desktop/projects/light-scattering/monte_carlo_cpp/src/PhaseFunctions.cpp#L11) uses HG with `g=0.7`, whose peak is much larger than 4. The acceptance-rejection sampler is therefore invalid.
  - Polarization is not implemented in the active MC transport. `p.stokes` is initialized at [MonteCarloDriver.cpp#L72](c:/Users/ashton/Desktop/projects/light-scattering/monte_carlo_cpp/src/MonteCarloDriver.cpp#L72), but no Mueller transform is ever applied and no Stokes output is accumulated.
  - Randomness is reproducible but not well designed for parallel transport. Each thread gets the same fixed seed at [MonteCarloDriver.cpp#L34](c:/Users/ashton/Desktop/projects/light-scattering/monte_carlo_cpp/src/MonteCarloDriver.cpp#L34), and surface reflection uses one shared global RNG at [SurfaceReflection.cpp#L7](c:/Users/ashton/Desktop/projects/light-scattering/monte_carlo_cpp/src/SurfaceReflection.cpp#L7).
  - Zenith binning is inconsistent. `zenDeg` is 0..180 at [MonteCarloDriver.cpp#L84](c:/Users/ashton/Desktop/projects/light-scattering/monte_carlo_cpp/src/MonteCarloDriver.cpp#L84), but `N_ZEN=90` at [MonteCarloDriver.cpp#L28](c:/Users/ashton/Desktop/projects/light-scattering/monte_carlo_cpp/src/MonteCarloDriver.cpp#L28), so anything above 89 degrees is clamped into the last bin at [MonteCarloDriver.cpp#L89](c:/Users/ashton/Desktop/projects/light-scattering/monte_carlo_cpp/src/MonteCarloDriver.cpp#L89).
- What is suspicious or incorrect, physics/modeling issues:
  - The estimator is the wrong quantity for your goal. `recordExitLocal()` bins photons escaping TOA by global direction at [MonteCarloDriver.cpp#L76](c:/Users/ashton/Desktop/projects/light-scattering/monte_carlo_cpp/src/MonteCarloDriver.cpp#L76), not radiance seen by a ground observer in a fisheye projection. There is no observer, no camera model, and no next-event estimator.
  - Free-path sampling uses the local extinction coefficient at the pre-step altitude at [MonteCarloDriver.cpp#L126](c:/Users/ashton/Desktop/projects/light-scattering/monte_carlo_cpp/src/MonteCarloDriver.cpp#L126) in a vertically varying medium. That is only an approximation, not a correct optical-depth sampling scheme for long slant paths in an exponential atmosphere.
  - The Rayleigh/aerosol mixture fraction is based on raw densities at the old altitude at [MonteCarloDriver.cpp#L135](c:/Users/ashton/Desktop/projects/light-scattering/monte_carlo_cpp/src/MonteCarloDriver.cpp#L135), not on the wavelength-dependent scattering coefficients actually used for transport.
  - Exit bins accumulate raw photon counts at [MonteCarloDriver.cpp#L94](c:/Users/ashton/Desktop/projects/light-scattering/monte_carlo_cpp/src/MonteCarloDriver.cpp#L94) with no solid-angle normalization, so even as a radiance histogram the bin values are not physically normalized.
  - The atmosphere is placeholder-only. [Atmosphere.cpp#L11](c:/Users/ashton/Desktop/projects/light-scattering/monte_carlo_cpp/src/Atmosphere.cpp#L11) and [Atmosphere.cpp#L21](c:/Users/ashton/Desktop/projects/light-scattering/monte_carlo_cpp/src/Atmosphere.cpp#L21) define only two layers up to 20 km, with zero density above that despite `R_TOA` at 100 km.
  - The spectrum is placeholder-only. [WavelengthHandling.cpp#L8](c:/Users/ashton/Desktop/projects/light-scattering/monte_carlo_cpp/src/WavelengthHandling.cpp#L8) creates 41 equal-weight bands with `fluxFraction=1.0` at [WavelengthHandling.cpp#L14](c:/Users/ashton/Desktop/projects/light-scattering/monte_carlo_cpp/src/WavelengthHandling.cpp#L14); that is not a solar spectrum.
- Overall MC conclusion: the active C++ Monte Carlo branch is not a correct twilight fisheye radiance solver. It is a debugging-stage transport prototype with serious geometry, estimator, and sampling errors. The older [spherical/MonteCarloSky.cpp](c:/Users/ashton/Desktop/projects/light-scattering/spherical/MonteCarloSky.cpp) is slightly cleaner as a toy transport demo, but it is also not observer-centered and has no polarization.
